```python
def validar_nombre(nombre):
    if (nombre.isalpha()):
        print (nombre)
    else:
        print("Error")

nombre= input("Ingrese un nombre:")
validar_nombre(nombre)

nombres=["Juan", "Ana", "Enrique 8", "Maria", "Luis"]

def validar_nombres(nombres):
    for nombre in nombres:
        validar_nombre(nombre)

validar_nombres(nombres)
```
