## 13. Reto 12 — Generar un resumen por clasificación

Implementa `generar_resumen(valores)` para contar cuántos elementos son `ALTO`, `NORMAL` y `BAJO`. Debe reutilizar `clasificar_valor()` y retornar los resultados.

Requisito: inicializa los contadores antes del ciclo.

````python
def generar_resumen(valores):
    altos = 0
    normales = 0
    bajos = 0

    for valor in valores:
        clasificacion = clasificar_valor(valor)
        if clasificacion == "ALTO":
            altos = altos + 1
        elif clasificacion == "NORMAL":
            normales = normales + 1
        else:
            bajos = bajos + 1

    return altos, normales, bajos

print(generar_resumen([22.5, 55.1, 72.3, 38.7, 61.4]))
````

## 14. Reto 13 — Añadir acumulador y bandera

Amplía el resumen para calcular también:
- La suma de todos los valores.
- Si existe al menos un valor clasificado como `ALTO`.

Usa una bandera inicializada en `False` y cámbiala a `True` cuando encuentres un valor `ALTO`. Retorna los contadores, la suma y la bandera.

````python
def generar_resumen_completo(valores):
    altos = 0
    normales = 0
    bajos = 0
    suma = 0
    existe_alto = False

    for valor in valores:
        suma = suma + valor
        clasificacion = clasificar_valor(valor)
        if clasificacion == "ALTO":
            altos = altos + 1
            existe_alto = True
        elif clasificacion == "NORMAL":
            normales = normales + 1
        else:
            bajos = bajos + 1

    return altos, normales, bajos, suma, existe_alto

print(generar_resumen_completo([22.5, 55.1, 72.3, 38.7, 61.4]))
````
