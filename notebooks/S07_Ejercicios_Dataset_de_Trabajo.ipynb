# 9. Reto 6 — Contador mediante una función

### Contexto

````python
valores = [22, 55, 72, 38, 61]
````

### Objetivo

Crear una función que reciba una lista y retorne la cantidad de valores
procesados.

````python
def contar_registros(valores):
    contador = 0
    for valor in valores:
        contador = contador + 1
    return contador
````

````python
print(contar_registros(valores))
````

### Resultado esperado


```text
5
```


### Restricción

Utilizar un contador.

### Pregunta

¿Por qué el contador debe inicializarse dentro de la función?

    Porque cada vez que llamas a la función debe empezar de cero. Si el contador estuviera afuera, conservaría el valor de la llamada anterior y los resultados saldrían mal. Por ejemplo, la primera llamada daría 5 y la segunda empezaría desde 5, dando 10 en vez de 5.

Al ponerlo dentro, la función siempre arranca limpia y da el resultado correcto, sin importar cuántas veces la uses. Además, así la función es independiente y no depende de variables externas.

   - `mostrar_resumen(total, altos, normales, bajos, suma, existe_alto)` solo muestra; recibe los seis datos por parámetro.
   - `ejecutar_opcion(opcion, valores, resultado)` ya no llama a `leer_opcion()`; recibe la opción y retorna el resultado actualizado.

> Nota: la guía propone `procesar_registros(cantidad)`. Como este proyecto procesa una lista, el parámetro pasa a ser `valores` y la cantidad se obtiene con `contar_registros()`.

La siguiente celda contiene la versión refactorizada (v0.7). Redefine las funciones de los retos anteriores, así que reemplaza las versiones con globales.

````python
# ===== Dataset de Trabajo v0.7 (refactorizado) =====

def clasificar_valor(valor):
    if valor > 70:
        clasificacion = "ALTO"
    elif valor >= 40:
        clasificacion = "NORMAL"
    else:
        clasificacion = "BAJO"
    return clasificacion


def contar_registros(valores):
    contador = 0
    for valor in valores:
        contador = contador + 1
    return contador


def calcular_suma(valores):
    suma = 0
    for valor in valores:
        suma = suma + valor
    return suma


def resumir_valores(valores):
    alto = 0
    normal = 0
    bajo = 0
    for valor in valores:
        categoria = clasificar_valor(valor)
        if categoria == "ALTO":
            alto = alto + 1
        elif categoria == "NORMAL":
            normal = normal + 1
        else:
            bajo = bajo + 1
    return alto, normal, bajo


def mostrar_menu():
    print("1. Procesar registros")
    print("2. Mostrar información")
    print("3. Salir")


def leer_opcion():
    opcion = input("Seleccione una opción: ")
    return opcion


def procesar_registros(valores):
    total = contar_registros(valores)
    altos, normales, bajos = resumir_valores(valores)
    suma = calcular_suma(valores)
    existe_alto = altos > 0
    return total, altos, normales, bajos, suma, existe_alto


def mostrar_resumen(total, altos, normales, bajos, suma, existe_alto):
    print("Resumen")
    print(f"Total: {total}")
    print(f"ALTO: {altos}")
    print(f"NORMAL: {normales}")
    print(f"BAJO: {bajos}")
    print(f"Suma: {suma}")
    print(f"¿Existe ALTO?: {existe_alto}")


def ejecutar_opcion(opcion, valores, resultado):
    """Coordina la opción. Retorna (continuar, resultado)."""
    match opcion:
        case "1":
            resultado = procesar_registros(valores)
            print("Registros procesados")
        case "2":
            if resultado is None:
                print("Primero procese los registros (opción 1)")
            else:
                mostrar_resumen(*resultado)
        case "3":
            print("Saliendo...")
            return False, resultado
        case _:
            print("Opción no válida")
    return True, resultado


def main():
    valores = [20, 55, 80, 40, 70]
    resultado = None
    continuar = True
    while continuar:
        mostrar_menu()
        opcion = leer_opcion()
        continuar, resultado = ejecutar_opcion(opcion, valores, resultado)
````

````python
main()
````

# 26. Preguntas de reflexión

1. ¿Qué problema resuelven las funciones?
> Las funciones nos permiten diseñar operaciones que pueden ser reutilizadas facilmente, haciendo que el codigo se vea más limpio y se pueda trabajar más eficientemente.
2. ¿Cuál es la diferencia entre un parámetro y un argumento?
> El parametro es una variable que se define en la función para que esta reciba información, mientras que un argumento es la información que le enviamos a la función.
3. ¿Por qué `return` es importante?
> El return es importante porque nos permite utilizar funciones y almacenar los valores que producen para poder utilizarlos en otra parte del codigo.
4. ¿Cuándo conviene utilizar una función?
> Conviene utilizar una funcion si se necesita realizar un proceso varias veces y/o en varias partes del codigo.
5. ¿Qué diferencia existe entre `print()` y `return`?
> El print simplemente muestra lo que le pedimos, mientras que el return guarda el valor y deja que se utilice en otra parte del codigo.
6. ¿Qué ventaja tiene dividir Dataset de Trabajo en funciones?
> Hace que el codigo sea más facil de entender y mantener, ademas de que si se requiere hacer un cambio, con una función solo se debe hacer el cambio en una parte del codigo, sin una funcion se tendria que hacer el cambio en multiples partes.
7. ¿Por qué no es conveniente depender innecesariamente de variables
   globales?
> Porque pueden generar dificultad a la hora de detectar y corregir errores o generar dependencias que son dificiles de comprender.
8. ¿Qué significa refactorizar?
> Refactorizar es organizar el codigo de una manera más limpia y facil de entender sin modificar el codigo en si, solo modificando la estructura.
9. ¿Qué riesgo existe al refactorizar un programa que ya funciona?
> Al refactorizar se pueden cometer errores y estos errores pueden afectar el comportamiento del codigo o directamente hacer que el codigo deje de funcionar.
10. ¿Por qué fue útil crear un commit antes de comenzar la refactorización?
> Porque nos permite tener la seguridad de que si se genera un error al refactorizar, podemos regresar facilmente a la version del codigo donde el commit fue hecho sin el riesgo de dañar permanentemente nuestro codigo.
